# Pump.fun MVP reproduction
This notebook runs the deterministic fixture path from a clean Python runtime. It does not make live requests or claim scientific observations.

```python
from pathlib import Path
import json, shutil, subprocess, sys
ROOT = Path.cwd()
if not (ROOT / 'run_mvp.py').exists():
    ROOT = next(p for p in [Path.cwd(), Path('/content/code')] if (p / 'run_mvp.py').exists())
OUT = ROOT / 'output' / 'notebook-reproduction'
if OUT.exists(): shutil.rmtree(OUT)
```

```python
result = subprocess.run([sys.executable, 'run_mvp.py', '--mode', 'fixture', '--output', str(OUT)], cwd=ROOT, text=True, capture_output=True)
print(result.stdout)
assert result.returncode == 0, result.stderr
```

```python
validation = json.loads((OUT / 'run' / 'validation_results.json').read_text())
manifest = json.loads((OUT / 'mvp_release' / 'release_manifest.json').read_text())
print('validation:', validation['status'])
print('events:', validation['event_count'])
print('release files:', len(manifest['files']))
assert validation['status'] == 'PASS'
```
